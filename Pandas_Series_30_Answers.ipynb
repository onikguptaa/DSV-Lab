## Question 27

Identify the NaN label, its `isnull` result, why it occurs, and labels with numeric values.

```python
d = {'Red': 12, 'Blue': 18, 'Green': 24}
labels = ['Green', 'Red', 'Yellow', 'Blue']
result = pd.Series(d, index=labels)
print(result)
print('isnull for Yellow:', result.isnull()['Yellow'])  # True
print('Numeric labels:', result[result.notnull()].index.tolist())
# Yellow is not a key in d, so its value becomes NaN.
```

## Question 28

Trace the dictionary conversion and state whether `s` and `s2` represent the same labelled data.

```python
s = pd.Series([90, 75, 82], index=['Ravi', 'Aman', 'Neha'])
d = s.to_dict()
s2 = pd.Series(d)
print('Dictionary:', d)
print('s2:')
print(s2)
print('Same labelled data:', s.to_dict() == s2.to_dict())
# The labels and values match; the order follows the dictionary's insertion order in modern Python.
```

## Question 29

Create a labelled Series for four products; show values/index, select a label, filter sales above 500, check a label, and convert to a dictionary.

```python
sales = pd.Series([650, 420, 800, 350], index=['Laptop', 'Mouse', 'Phone', 'Keyboard'])
print('Values:', sales.values)
print('Index:', sales.index)
print('Phone sales:', sales['Phone'])
print('Sales above 500:')
print(sales[sales > 500])
print('Mouse exists:', 'Mouse' in sales.index)
print('Dictionary:', sales.to_dict())
```

## Question 30

Analyse index order, NaN, filter above 20, `notnull`, Series name and index name.

```python
data = {'A': 15, 'B': 35, 'C': 25}
s = pd.Series(data, index=['C', 'A', 'D', 'B'])
s.name = 'Observations'
s.index.name = 'Category'
print(s)
print('Index order:', list(s.index))  # C, A, D, B
print('Missing labels:', s[s.isnull()].index.tolist())  # D
print('Values > 20:')
print(s[s > 20])  # C=25, B=35
print('Not null:')
print(s.notnull())
print('Series name:', s.name)
print('Index name:', s.index.name)
```
